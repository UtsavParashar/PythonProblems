### Iterators and Generators

````python
### Iterators and Generators
### Iterators - iter() -> return iterator object + next() -> checks and returns StopIteration Ex whenever required
class Counter:
    def __init__(self, limit):
        self.limit = limit
        self.curr = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.curr >= self.limit:
            raise StopIteration

        val = self.curr
        self.curr += 1
        return val

counter = Counter(3)
for x in counter:
    print(x)
````

````python
# Stack Frame, Heap Frame
# Function live and die in stack frame, all elements for a function are loaded in memory under stack frame and then it dies at return.
# Generator - yield let's compiler know to generate the object in heap frame so that it could live and give control back to the calling object.
# Generator also uses iterator, the class above could be written in generator as

# Generator creates a generator object. The generator contains suspended execution state.

# return says: "I'm done. Give the caller this result."
# yield says: "Give the caller this result, but pause me here. I may continue later."

def counter(limit):
    for i in range(limit):
        yield i

for x in counter(3):
    print(x)
````

````python
# Generators allow lazy production of values.
# Lazy means: Don't compute something until it is actually needed.
# Generators can reduce memory consumption and enable streaming/lazy computation. Performance depends on the workload.
def counter():
    yield 0
    yield 1
    yield 2

g = counter() # g points to the generator object
print(next(g))
print(next(g))
print(next(g))
````

#### Descriptors
- Descriptors allow an object to control attribute access.  

A descriptor is an object that defines one or more of:
``__get__``
``__set__``
``__delete__``

Think of a Descriptor as a smart middleman sitting between you and an object's internal dictionary.

A Python object becomes a Descriptor simply by defining at least one of these three magic methods:

__get__(self, instance, owner): Defines what happens when the attribute is read.

__set__(self, instance, value): Defines what happens when the attribute is written.

__delete__(self, instance): Defines what happens when the attribute is deleted.

````python
# why do we need descriptors?

class User:
    def __init__(self, age):
        if age < 0:
            raise ValueError
        self.age = age

# Every age must be between 0 and 150.
# You could write validation in __init__.
#But what if age can be changed later?

user = User(100)
print(user.age)
user.age = -120
print(user.age)

# Descriptors let you centralize attribute behavior.
````

````python
# Descriptor
class PositiveInteger:

    def __set_name__(self, owner, name):
        # Called automatically when the class is created.
        # Saves the attribute name (e.g., 'age' or 'height').
        self.private_name = "_" + name

    def __get__(self, instance, owner):
        if instance is None:
            return self  # Accessed directly via class (e.g., User.age)
        return getattr(instance, self.private_name, None)

    def __set__(self, instance, value):
        if not isinstance(value, int) or value < 0:
            raise ValueError(f"Value must be a non-negative integer.")

        print(dir(instance), value)
        setattr(instance, self.private_name, value)

class User:
    # We attach instances of the Descriptor as CLASS attributes:
    age = PositiveInteger()
    height = PositiveInteger()

    def __init__(self, age, height):
        self.age = age  # Triggers PositiveInteger.__set__!
        self.height = height


# --- Usage ---
u = User(25, 175)
print(u.age)  # 25 (Triggers PositiveInteger.__get__)

u.age = -10  # Raises ValueError: Value must be a non-negative integer.
````

- A test is an executable specification of expected behavior.
- Testing is - Arrange - Act - Assert
- Arrange - Prepare the state
- Act - Execute the behavior
- Assert - Verify the result

````python
def add(a, b):
    return a+b

def test_add():
    # Arrange
    a = 2
    b = 3

    # Act
    result = add(a, b)

    # Assert
    assert result == 5

test_add()
````

- Unit - Small + isolated + fast

#### Subprocess
- A process is an operating-system execution unit.
- The parent process can communicate with child process using stdin, stdout, stderr and exit code
- result = subprocess.run() can be used
- result.stdout...
- Why use a list instead of a shell command string? The list form avoids unnecessary shell interpretation., This becomes extremely important for security.
- `shell=True` - shell=True asks the system to execute the command through a shell. That enables shell features, but also increases injection risk when inputs are untrusted. Use it only when shell semantics are actually required, and never interpolate untrusted input into a shell command.
- Prefer argument lists:  `subprocess.run(["git", "status"])` over `subprocess.run("git status", shell=True)`
- subprocess.run() is convenient when you mostly want:-> start -> wait -> collect result
- But what if you need ongoing interaction? -> Start Compiler -> Send Input -> Read Output -> Send another command, you can use
- `subprocess.Popen(...)` - This gives you more control over the process lifecycle and its pipes.
- Popen ->  +-- stdin - +-- stdout
 +-- stderr
 +-- poll()
 +-- wait()
 +-- terminate()
 +-- kill()

##### Static analysis - Analyzing a program without executing it. 
##### Static analysis vs dynamic analysis 
- Static -> Source Code - Parser - analysis - result -> No execution required
- Dynamic -> Source Code - execution - behavior
- Static analysis is generally safer for inspecting arbitrary code because you don't need to execute it. But static analysis cannot know everything.

##### AST vs CST
- If you're building a code formatter or refactoring tool that needs precise source representation, a Concrete Syntax Tree (CST) or equivalent concrete representation can be more appropriate.
- AST → "What does this code mean structurally?"
- CST → "How exactly was this code written?"
- AST could be used for for dead-code detection
- AST for duplicate-code detection
- The AST sits between source representation and executable representation.
