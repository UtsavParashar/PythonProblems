### Closure and Decorator

````python
## variable scoping is LEGB - Local, Enclosed, Global, Builtin
### Closures are nested functions where the inner function has acccess to the outer functions 
### variables(enclosed) even after the outer function execution is complete.
````

````python
## Higher Order function - function which accepts and/or returns a function

## closure - the inner function wrapper when pointed by hello variable also has 
# access to x variable of outer function is called closure as x in enclosed scope for inner function

## decorator - decorates a function with additional capability


def decorator(func):
    print("decorating")

    def wrapper():
        print("before")
        func()
        print("after")

    return wrapper


# @decorator
def hello():
    print("hello")

hello = decorator(hello)
print('middle')
hello()
````

````python
import functools
def decorator(func):
    print("decorating")
    x = 10
    @functools.wraps(func) # logging as per the decorated func
    def wrapper(*args, **kwargs):
        print("before")
        print(f'{x}, {args[0]}')
        func('atm')
        print("after")

    return wrapper

# @decorator(3) == hello = parent_decorator(3, decorator(hello))
# @decorator2 == hello = decorator2(decorator(hello))
# @decorator == hello = decorator(hello)
def hello(param):
    print("hello", param)

hello = decorator(hello)
print('middle')
hello('ctm')
````

### Iterators and Generators

````python
### Iterators and Generators
### Iterators - iter() -> return iterator object + next() -> checks and returns StopIteration Ex whenever required
class Counter:
    def __init__(self, limit):
        self.limit = limit
        self.curr = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.curr >= self.limit:
            raise StopIteration

        val = self.curr
        self.curr += 1
        return val

counter = Counter(3)
for x in counter:
    print(x)
````

````python
# Stack Frame, Heap Frame
# Function live and die in stack frame, all elements for a function are loaded in memory under stack frame and then it dies at return.
# Generator - yield let's compiler know to generate the object in heap frame so that it could live and give control back to the calling object.
# Generator also uses iterator, the class above could be written in generator as

# Generator creates a generator object. The generator contains suspended execution state.

# return says: "I'm done. Give the caller this result."
# yield says: "Give the caller this result, but pause me here. I may continue later."

def counter(limit):
    for i in range(limit):
        yield i

for x in counter(3):
    print(x)
````

````python
# Generators allow lazy production of values.
# Lazy means: Don't compute something until it is actually needed.
# Generators can reduce memory consumption and enable streaming/lazy computation. Performance depends on the workload.
def counter():
    yield 0
    yield 1
    yield 2

g = counter() # g points to the generator object
print(next(g))
print(next(g))
print(next(g))
````

#### Inheritance
##### MRO - Method Resolution Order

````python
class Animal:
    def speak(self):
        print('speak')
        return '---'

class Dog(Animal):
    def bark(self):
        print('bark')
        return '==='

print(f'Animal.__dict__ - {Animal.__dict__}')
print(f'Dog.__dict__ - {Dog.__dict__}')

dog = Dog()
print(dog.bark())
print(dog.speak())

print(f'Animal mro -- {Animal.__mro__}')
print(f'Dog mro -- {Dog.__mro__}')
print(f'dog mro -- {dog.__mro__}')
````

````python
### Diamond Problem

class A:
    def hello(self):
        print("A")

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

print(f'D MRO - {D.__mro__}')
````

#### Descriptors
- Descriptors allow an object to control attribute access.  

A descriptor is an object that defines one or more of:
``__get__``
``__set__``
``__delete__``

Think of a Descriptor as a smart middleman sitting between you and an object's internal dictionary.

A Python object becomes a Descriptor simply by defining at least one of these three magic methods:

__get__(self, instance, owner): Defines what happens when the attribute is read.

__set__(self, instance, value): Defines what happens when the attribute is written.

__delete__(self, instance): Defines what happens when the attribute is deleted.

````python
# why do we need descriptors?

class User:
    def __init__(self, age):
        if age < 0:
            raise ValueError
        self.age = age

# Every age must be between 0 and 150.
# You could write validation in __init__.
#But what if age can be changed later?

user = User(100)
print(user.age)
user.age = -120
print(user.age)

# Descriptors let you centralize attribute behavior.
````

````python
# Descriptor
class PositiveInteger:

    def __set_name__(self, owner, name):
        # Called automatically when the class is created.
        # Saves the attribute name (e.g., 'age' or 'height').
        self.private_name = "_" + name

    def __get__(self, instance, owner):
        if instance is None:
            return self  # Accessed directly via class (e.g., User.age)
        return getattr(instance, self.private_name, None)

    def __set__(self, instance, value):
        if not isinstance(value, int) or value < 0:
            raise ValueError(f"Value must be a non-negative integer.")

        print(dir(instance), value)
        setattr(instance, self.private_name, value)

class User:
    # We attach instances of the Descriptor as CLASS attributes:
    age = PositiveInteger()
    height = PositiveInteger()

    def __init__(self, age, height):
        self.age = age  # Triggers PositiveInteger.__set__!
        self.height = height


# --- Usage ---
u = User(25, 175)
print(u.age)  # 25 (Triggers PositiveInteger.__get__)

u.age = -10  # Raises ValueError: Value must be a non-negative integer.
````

````python
class Instrument:
    asset_class = "Unknown"

    def __init__(self, symbol):
        self.symbol = symbol

    def __repr__(self):
        return f"{type(self).__name__}(symbol={self.symbol!r})"

    def __eq__(self, other):
        if not isinstance(other, Instrument):
            return NotImplemented
        return self.symbol == other.symbol


class Equity(Instrument):
    asset_class = "Equity"

    def price(self):
        return 100

e = Equity("AAPL")
e.asset_class # Equity class object has asset_class and not e instance object

e.price() # Equity.price(e)
Equity.price(e)

e.__repr__()

a = Equity("AAPL")
b = Equity("AAPL")
a is b
a == b
````
