### Closure and Decorator

```python
## variable scoping is LEGB - Local, Enclosed, Global, Builtin
### Closures are nested functions where the inner function has acccess to the outer functions 
### variables(enclosed) even after the outer function execution is complete.
```

```python
## Higher Order function - function which accepts and/or returns a function

## closure - the inner function wrapper when pointed by hello variable also has 
# access to x variable of outer function is called closure as x in enclosed scope for inner function

## decorator - decorates a function with additional capability


def decorator(func):
    print("decorating")

    def wrapper():
        print("before")
        func()
        print("after")

    return wrapper


# @decorator
def hello():
    print("hello")

hello = decorator(hello)
print('middle')
hello()
```

```python
import functools
def decorator(func):
    print("decorating")
    x = 10
    @functools.wraps(func) # logging as per the decorated func
    def wrapper(*args, **kwargs):
        print("before")
        print(f'{x}, {args[0]}')
        func('atm')
        print("after")

    return wrapper

# @decorator(3) == hello = parent_decorator(3, decorator(hello))
# @decorator2 == hello = decorator2(decorator(hello))
# @decorator == hello = decorator(hello)
def hello(param):
    print("hello", param)

hello = decorator(hello)
print('middle')
hello('ctm')
```

### Iterators and Generators

```python
### Iterators and Generators
### Iterators - iter() -> return iterator object + next() -> checks and returns StopIteration Ex whenever required
class Counter:
    def __init__(self, limit):
        self.limit = limit
        self.curr = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.curr >= self.limit:
            raise StopIteration

        val = self.curr
        self.curr += 1
        return val

counter = Counter(3)
for x in counter:
    print(x)
```

```python
# Stack Frame, Heap Frame
# Function live and die in stack frame, all elements for a function are loaded in memory under stack frame and then it dies at return.
# Generator - yield let's compiler know to generate the object in heap frame so that it could live and give control back to the calling object.
# Generator also uses iterator, the class above could be written in generator as

# Generator creates a generator object. The generator contains suspended execution state.

# return says: "I'm done. Give the caller this result."
# yield says: "Give the caller this result, but pause me here. I may continue later."

def counter(limit):
    for i in range(limit):
        yield i

for x in counter(3):
    print(x)
```

```python
# Generators allow lazy production of values.
# Lazy means: Don't compute something until it is actually needed.
# Generators can reduce memory consumption and enable streaming/lazy computation. Performance depends on the workload.
def counter():
    yield 0
    yield 1
    yield 2

g = counter() # g points to the generator object
print(next(g))
print(next(g))
print(next(g))
```

```python
# List comprehension vs Generator Expression
data = range(5)
[x * x for x in data] # Give me all answers now
(x * x for x in data) # Give me a mechnanism for producing the answers when requested.
{x * 2 for x in data} # Set Comprehension
{x: x * x for x in data} # Dict Comprehension
```

### Class vs Instance

#### Class - A class is an object that defines behavior and is used to create/manage instances.

```python
class Trade:
    pass

t1 = Trade()
t2 = Trade()

# There are now three important objects:
# Trade is the class. - Class Object
# t1 and t2 are instances - Instance Objects
# The class describes the structure/behavior shared by its instances.
```

```python
class Trade:
    asset_class = 'Equity'
    def __init__(self, sym, px):
        self.sym = sym
        self.px = px

    def value(self):
        return self.px

t1 = Trade('AAPL', 100)
print(t1.value())

print(Trade.value)
print(t1.value)
print(t1.__dict__) # Instance variables are stored as dict
print(Trade.__dict__)
print(Trade.__init__)
```

##### __dict__ -> dict dunder contains object's instanc attributes.

```python
class Trade:
    asset_class = 'Equity'

    def __init__(self, sym):
        self.sym = sym

t1 = Trade('AAPL')
t2 = Trade('GOOG')

print(f'Trade class dict - {Trade.__dict__}')
print(f't1 dict - {t1.__dict__}')
print(f't2 dict - {t2.__dict__}')

print(f't1 asset class - {t1.asset_class}')
print(f't2 asset class - {t2.asset_class}')

t1.asset_class = 'Fixed Income'
print(f't1 asset class - {t1.asset_class}')
print(f't1 dict - {t1.__dict__}')
```

```python
### Dunders
## __dict__, __init__, __new__, __iter__, __next__, __repr__, __str__, __lt__, __gt__, __eq__, __hash__

## Operator Overloading
class Money:
    def __init__(self, amt):
        self.amt = amt

    def __add__(self, other):
        return Money(self.amt + other.amt)

a = Money(100)
b = Money(50)
c = a + b
print(c)
print(a.__add__(b))
```

```python
# __repr__(self) -> is intended to be an informative/developer-oriented representation.
# __str__(self) -> is intended for user/human friendly representation.

class Trade:
    def __init__(self, symbol, price):
        self.symbol = symbol
        self.price = price

    def __repr__(self):
        return f"Trade(symbol={self.symbol!r}, price={self.price})"

    def __str__(self):
        return f"{self.symbol} @ {self.price}"
```

#### Inheritance
##### MRO - Method Resolution Order

```python
class Animal:
    def speak(self):
        print('speak')
        return '---'

class Dog(Animal):
    def bark(self):
        print('bark')
        return '==='

print(f'Animal.__dict__ - {Animal.__dict__}')
print(f'Dog.__dict__ - {Dog.__dict__}')

dog = Dog()
print(dog.bark())
print(dog.speak())

print(f'Animal mro -- {Animal.__mro__}')
print(f'Dog mro -- {Dog.__mro__}')
print(f'dog mro -- {dog.__mro__}')
```

```python
### Diamond Problem

class A:
    def hello(self):
        print("A")

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

print(f'D MRO - {D.__mro__}')
```

```python
#### super() - super() provides a proxy that continues attribute/method lookup according to the MRO from the current class.

class A:
    def hello(self):
        print("A")

class B(A):
    def hello(self):
        print("B")
        super().hello()

B().hello()
```

```python
class A:
    def hello(self):
        print("A")

class B(A):
    def hello(self):
        print("B")
        super().hello()

class C(A):
    def hello(self):
        print("C")
        super().hello()

class D(B, C):
    def hello(self):
        print("D")
        super().hello()

D().hello()
```
