#### The key concept: parameters/weights

- Suppose we have 
```
Input                Output

2                    4
3                    6
4                    8
5                   10
```
- We want a machine to discover: `output = 2 × input`
- We could represent this as: `y = wx`
```
where:

x = input
y = predicted output
w = parameter
```
- Initially say w = 1.5 then for x = 5, y would be 7.5 but that is the wrong answer. It's an error.
- `So the machine adjusts w, and eventually w == 2 and then y == 10` - `That's Machine Learning`
- `So - The parameter/weight determines how the model transforms input into output.`
- `An LLM simply has an enormous number of these parameters arranged across many layers.`
- OpenAI describes modern ML models similarly: they consist of large collections of numerical parameters/weights, and training adjusts those values to reflect patterns learned from training data.

##### How does the model know that it is wrong? - LOSS

- Suppose, 
```
Actual = 10
Predicted = 6
```
- We need a numerical measurement of "how bad was that prediction?"
- For eg. `Loss = (Actual - Prediction)^2`
- Therefore - `Loss = (10-6)^2 == 16`
- If prediction were - 9, then Loss = 1
```
Prediction     Loss

6              16
9               1
10              0
```
#### Loss is a numerical measure of how badly the model performed on its training objective.
- For language models, the commonly used objective is based on predicting the correct next token and measuring the probability assigned to that token; cross-entropy is a central loss used for language-model training. Scaling-law research, for example, analyzes language-model performance using `cross-entropy loss`.

- Now we have
```
Input
  ↓
Model
  ↓
Prediction
  ↓
Compare with correct answer
  ↓
Loss
```
- `How does the model improve? Gradient Descent`


#### Gradient descent — first principles
- Imagine you're standing on a mountain.
- Your objective: Reach the bottom.
- You can't see the entire landscape.
- You can only determine: Which direction slopes downward from where I currently stand?
- You take a small step downhill. Then repeat. That's the intuition behind gradient descent.

##### Gradient descent - Analogy
- Imagine adjusting the temperature of an oven.
- You don't know the perfect temperature.
- You bake something.
- It's too cold.
- You increase temperature.
- Still too cold.
- Increase again.
- Eventually: `too cold → better → optimal`
- The model does something analogous
```
bad parameters
      ↓
make prediction
      ↓
measure error
      ↓
determine which parameter changes reduce error
      ↓
change parameters
      ↓
repeat
```

-Imagine
```
Parameter w
     ↓
Prediction
     ↓
Loss
```
- We want - ` Loss ↓`
- So we ask - `If I slightly increase w, does loss increase or decrease?`
- mathematically, we calculate a derivative - dLoss/dw
- `How sensitive is the loss to this parameter?`
- Then we update
```
wnew ​= wold ​− η ∂w/∂Loss​
where:

η = learning rate
derivative = direction/sensitivity
subtract = move toward lower loss
```



#### Neural networks

#### Where does LLM pretraining fit?

- At first glance, you might think:
- `"Who labels billions of sentences as the correct next word?"`
- Nobody needs to manually label them.
- Eg. `The cat sat on the mat.`
- We can automatically create training examples:
```
"The"                   → "cat"
"The cat"               → "sat"
"The cat sat"           → "on"
"The cat sat on"        → "the"
"The cat sat on the"    → "mat"
```
- The text itself provides the supervision.
- This is often called self-supervised learning.
- The training signal is generated from the data itself.
- That is one of the most important ideas behind modern LLMs.

##### What changes inside an LLM when it learns?
- During training, the model repeatedly makes predictions from training examples, calculates a loss measuring how far the prediction is from the desired target, computes gradients showing how the loss changes with respect to each parameter, and uses an optimizer such as gradient descent to update the model's weights. Learning is therefore fundamentally the repeated adjustment of numerical parameters so that the model becomes better at its training objective.

##### Why can't a neural network directly understand text?
- Consider - `"I like cats"`
- A neural network operates on numbers.
- It doesn't directly operate on:
```
"I"
"like"
"cats"
```
- We need to convert language into numbers.
- The first step is: `Tokenization`

##### What is a token?
- A token is a unit used by the model's tokenizer.
- It might correspond to:
```
a word
a subword
a punctuation mark
part of a word
```
- For example, conceptually: `"unbelievable"` might be broken down into something like
```
"un"
"believ"
"able"
```
- `The exact tokenization depends on the tokenizer/model.`
- Therefore, don't say: "One token equals one word." That's wrong.
- A better statement: `A token is a model-specific unit of text used as the basic element for language modeling.`

##### Token IDs - Enumeration
- The model's vocabulary might conceptually look like:
```
Token        ID

the          100
cat          2048
dog          3712
sat          918
```

```
Therefore: "The cat sat"
might become: [100, 2048, 918]
These IDs are just integers.
- `The integer 2048 doesn't inherently mean anything mathematically.`
It is an index into the vocabulary.
```

##### Why can't we just use token IDs?
- Suppose:
```
cat = 100
car = 101
dog = 102
```
- If we feed these directly to a neural network, it might incorrectly interpret:`cat is numerically closer to car than dog` 
- because 101-100 = 1 and 102-100 = 2
- But these numerical distances are meaningless.
- We need a representation where relationships can be learned. `Enters Embeddings`.

#### Emneddings
- Instead of: `cat -> 100`
- we map the token to a vector - `cat → [0.21, -0.73, 0.44, ...]`
- maybe - `dog -> [0.19, -0.70, 0.47, ...]`
- and `car → [-0.82, 0.31, -0.12, ...]`
- Imagine every word/token is a point in a giant multidimensional space.
```
              dog
             ●

        cat ●

                      car ●
```
- Tokens with related usage patterns tend to develop related representations.
```
king
queen
prince
princess
```
may occupy related regions.
- But don't overinterpret embeddings as dictionaries.
- An embedding isn't:
```
dimension 1 = animal
dimension 2 = gender
dimension 3 = size
```
- Real neural representations are distributed and much more complicated.

##### Where do embeddings come from?
- This is another important first-principles question.
- We don't manually create them.
- The embedding values are learned.
- Conceptually:
```
token ID
   ↓
embedding lookup
   ↓
vector
```
- The embedding table itself contains learned parameters.
```
Vocabulary size = V
Embedding dimension = d
```
- Then the embedding matrix is roughly:E ∈ R^(V×d)
- If
```
V = 100,000
d = 4,096
```
- then the embedding matrix contains: `100,000 × 4,096 learned Numbers`.

##### What does the model actually predict?
- Suppose - `"The capital of France is"`
- The model might produce scores like - 
```
Paris       15.2
London       5.1
Berlin       3.4
Madrid       2.8
banana      -4.3
```
- These raw scores are called `Logits`. These are not probabilities yet.

##### From logits to probabilities
- We need to convert scores into a probability distribution. `That's what softmax does.`
```
logits
 ↓
softmax(Multiclass choices)/Sigmoid(Binary Choices)
 ↓
probabilities
```
```
Paris      0.94
London     0.02
Berlin     0.01
Madrid     0.01
...
```
- The probabilities sum to 1.
- So the LLM has effectively answered: `Given the context, how plausible is each possible next token?`

##### Softmax Function -
- The softmax function is a mathematical formula used in machine learning to turn a vector of raw scores (logits) into a probability distribution that sums up to 1 (or 100%).
- It is almost always used as the final activation layer in a deep learning model for multi-class classification tasks (e.g., identifying if an image is a cat, dog, or bird).

##### The fundamental equation of an autoregressive language model
- Conceptually: `P(xt​∣x1​,x2​,…,xt−1​) -> Probability of the next token given all previous tokens.`
- Eg:
```
P(
    "Paris" | "The capital of France is"
)
```
- The model estimates that probability.
- This is the fundamental engine behind autoregressive text generation.

- `The capital of France is Paris.` -> It isn't necessarily generating the entire sentence in one operation.
- Conceptually
```
"The capital of France is"
             ↓
          predict
             ↓
           Paris
             ↓
"The capital of France is Paris"
             ↓
          predict
             ↓
             .
             and so on
```
-`Text generation is fundamentally an iterative next-token prediction process.`
- OpenAI's discussion of hallucinations explicitly describes pretrained language models as learning through next-word prediction over huge quantities of text.

##### Why can next-token prediction become so powerful?
- This is the deeper question.
- Imagine training on:`Millions/billions of documents`
- To predict: `next token`
- the model cannot succeed merely by memorizing local word frequencies.
- It must learn patterns such as:
```
syntax
semantics -> Make sense of sentence (differenciate between gibbrish and a meaningful sentence)
grammar
facts
style
code structure
mathematical patterns
relationships
long-range dependencies
```
- Eg - `"The company reported earnings of $5 billion, which..."`
- To predict the next token well, the model benefits from understanding financial language.
- Likewise: `def calculate_vwap(prices, volumes):`
- To predict what comes next, the model needs patterns from programming code.
- This is why a seemingly simple objective can produce sophisticated capabilities.

##### But does the model "understand"?
- An LLM is trained with a token-prediction objective, but the internal representations learned to optimize that objective can encode rich statistical and semantic structure, enabling capabilities that look like reasoning, abstraction, translation, coding and planning. Exactly how these capabilities emerge remains an active area of research.

##### Why do we need tokenization?
```
Neural networks operate on numerical representations, not raw strings. Tokenization converts text into discrete vocabulary elements, and each token is mapped to an integer ID. Those IDs are then used to retrieve learned vector representations called embeddings, which provide the numerical input to the neural network.
```

##### What is the difference between a token ID and an embedding?
```
Token ID
↓
integer index

Embedding
↓
learned vector representation

```
- Eg - 
```
cat
 ↓
ID = 2048
 ↓
embedding = [0.12, -0.43, 0.91, ...]
```
- The ID is a lookup key.
- The embedding is the learned representation.

#### Temperature
- Now suppose probabilities are:
```
Paris       0.60
London      0.20
Berlin      0.10
Madrid      0.10
```
- There are multiple possible outputs.
- Temperature modifies how sharply concentrated the distribution is.
Conceptually:
```
Low temperature
↓
more deterministic

High temperature
↓
more randomness/diversity
```
- Temperature ≈ "How adventurous should sampling be?"
```
Low: More deterministic output
Paris: 0.95
London: 0.02

High: More creative outputs

Paris: 0.45
London: 0.25
Berlin: 0.15
...
```
The exact mathematical implementation depends on the model/API, but conceptually temperature changes the sharpness of the token distribution before sampling.

##### Sampling
Suppose:
```
Paris       0.70
London      0.20
Berlin      0.10
```
- The model doesn't necessarily have to always choose: `argmax(probability)`
- which would always give: `Paris`.

- Instead, it can sample from the distribution.
- Possible outcomes:
```
70% Paris
20% London
10% Berlin
```
`This introduces stochasticity.`

##### Temperature vs sampling
##### Sampling - Choose a token according to some probability distribution.
##### Temperature - One mechanism for modifying the distribution before sampling.
```
model
 ↓
logits
 ↓
temperature transformation
 ↓
probability distribution
 ↓
sampling
 ↓
next token
```
