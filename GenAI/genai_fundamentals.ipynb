#### AI → Machine Learning
- AI means building systems that perform tasks that we associate with intelligence. Eg. Recognize Image, recommend movie etc
- Historically, one way to build AI was to explicitly program rules.
```
if temperature > 30:
    print("hot")
else:
    print("not hot")
```
- You wrote the intelligence.
- The problem?
- Real-world problems have enormous numbers of rules.
- Consider recognizing a cat. You could theoretically write:
```
if ears == ...
and eyes == ...
and fur == ...
and shape == ...
...
```
- But this becomes impossible.
- So we move from:
- "Tell the computer the rules."
- to:
- "Give the computer examples and let it learn the rules."
`That's Machine Learning.`

#### The key concept: parameters/weights

- Suppose we have 
```
Input                Output

2                    4
3                    6
4                    8
5                   10
```
- We want a machine to discover: `output = 2 × input`
- We could represent this as: `y = wx`
```
where:

x = input
y = predicted output
w = parameter
```
- Initially say w = 1.5 then for x = 5, y would be 7.5 but that is the wrong answer. It's an error.
- `So the machine adjusts w, and eventually w == 2 and then y == 10` - `That's Machine Learning`
- `So - The parameter/weight determines how the model transforms input into output.`
- `An LLM simply has an enormous number of these parameters arranged across many layers.`
- OpenAI describes modern ML models similarly: they consist of large collections of numerical parameters/weights, and training adjusts those values to reflect patterns learned from training data.

##### How does the model know that it is wrong? - LOSS

- Suppose, 
```
Actual = 10
Predicted = 6
```
- We need a numerical measurement of "how bad was that prediction?"
- For eg. `Loss = (Actual - Prediction)^2`
- Therefore - `Loss = (10-6)^2 == 16`
- If prediction were - 9, then Loss = 1
```
Prediction     Loss

6              16
9               1
10              0
```
#### Loss is a numerical measure of how badly the model performed on its training objective.
- For language models, the commonly used objective is based on predicting the correct next token and measuring the probability assigned to that token; cross-entropy is a central loss used for language-model training. Scaling-law research, for example, analyzes language-model performance using `cross-entropy loss`.

- Now we have
```
Input
  ↓
Model
  ↓
Prediction
  ↓
Compare with correct answer
  ↓
Loss
```
- `How does the model improve? Gradient Descent`


#### Gradient descent — first principles
- Imagine you're standing on a mountain.
- Your objective: Reach the bottom.
- You can't see the entire landscape.
- You can only determine: Which direction slopes downward from where I currently stand?
- You take a small step downhill. Then repeat. That's the intuition behind gradient descent.

##### Gradient descent - Analogy
- Imagine adjusting the temperature of an oven.
- You don't know the perfect temperature.
- You bake something.
- It's too cold.
- You increase temperature.
- Still too cold.
- Increase again.
- Eventually: `too cold → better → optimal`
- The model does something analogous
```
bad parameters
      ↓
make prediction
      ↓
measure error
      ↓
determine which parameter changes reduce error
      ↓
change parameters
      ↓
repeat
```

-Imagine
```
Parameter w
     ↓
Prediction
     ↓
Loss
```
- We want - ` Loss ↓`
- So we ask - `If I slightly increase w, does loss increase or decrease?`
- mathematically, we calculate a derivative - dLoss/dw
- `How sensitive is the loss to this parameter?`
- Then we update
```
wnew ​= wold ​− η ∂w/∂Loss​
where:

η = learning rate
derivative = direction/sensitivity
subtract = move toward lower loss
```
