#### 1974. Minimum Time to Type Word Using Special Typewriter

* https://leetcode.com/problems/minimum-time-to-type-word-using-special-typewriter/description/


```python
class Solution:
    def minTimeToType(self, word: str) -> int:
        curr_char = 'a'
        res = 0
        for target_char in word:
            straight_dist = abs(ord(target_char) - ord(curr_char))
            # in circle if straight distance between a to b is b - a then circular dist is total - (b-a)
            circular_dist = 26 - straight_dist 

            d = min(straight_dist, circular_dist)

            res += d + 1
            curr_char = target_char
        
        return res
```
