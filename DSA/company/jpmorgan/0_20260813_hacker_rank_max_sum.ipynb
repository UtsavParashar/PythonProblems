* Return the max sum from the array * (index + 1) in which you are allowed to interchange element with its next element.
* Eg. [9, 7, 3]

```python
# it can be solved using 1d dp, similar to house robber

def max_res(nums):
    n = len(nums)
    p2, p1 = 0, (nums[-1] * n)
    for i in range(n-2, -1, -1):
        no_swap = (nums[i] * (i+1)) + (p1)
        swap = (nums[i+1] *(i+1)) + (nums[i]*(i+2)) + p2
        cr = max(no_swap, swap)

        p2, p1 = p1, cr

    return p1

max_res([9, 7, 3])
```
