#### 1446. Consecutive Characters
* https://leetcode.com/problems/consecutive-characters/description/

```python
class Solution:
    def maxPower(self, s: str) -> int:
        n = len(s)
        if n <= 0:
            return 0
        
        global_max = 1
        curr = 1
        for i in range(1, n):
            if s[i] == s[i-1]:
                curr += 1
                if curr > global_max:
                    global_max = curr
            else:
                curr = 1
        
        return global_max

Solution().maxPower('abbcb')
Solution().maxPower('aaabb')
Solution().maxPower('')
Solution().maxPower('aaaaa')
Solution().maxPower('abbbb')
```

# Extended the problem to return the char as well

```python
def get_max(s):
    if len(s) <= 0:
        return 0, ''
    
    n = len(s)
    global_max = 1
    curr = 1
    max_char = s[0]
    for i in range(1, n):
        if s[i] == s[i-1]:
            curr += 1
        else:
            curr = 1

        if curr > global_max:
            global_max = curr
            max_char = s[i]

    return global_max, max_char

#print(get_max('abbcb'))
#get_max('aaabb')
#get_max('')
#get_max('aaaaa')
#get_max('abbbb')
get_max('ccccd')
```

### Write a peformance counter of a function

```python
import time 


def get_account():
    time.sleep(6)
    return None


def get_user(id):
    time.sleep(6)
    raise ValueError('Test')
    return id

def performance_counter(func, *args):
    t = 5
    st = time.perf_counter()
    try:
        func_res = func(*args)
        if func_res:
            print(f'Function Res - {func.__name__}{func_res}')
    except Exception as e:
        #raise e
        print(e)
    finally:
        et = time.perf_counter()
        time_taken = et - st
        if time_taken > 5:
            print(f'function {func.__name__} took more than threshold of {t} seconds, time taken - {time_taken}')
        


performance_counter(get_user, 1)
performance_counter(get_account)
```
